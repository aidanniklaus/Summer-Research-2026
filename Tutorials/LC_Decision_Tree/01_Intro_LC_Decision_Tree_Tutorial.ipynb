## Step 1: Choose What to Download from MAST

#### 1.1: Pause... What Exactly is MAST??
The [Mikulski Archive for Space Telescopes (MAST)](https://archive.stsci.edu/) is a repository where data from publicly-funded astrophysics missions is hosted. It is free to use and all we need to access our target star is some sort of identification. In these tutorials, we will use TESS Input Catalog (TIC) numbers, but you can equally use a star's name or coordinates.

#### 1.2: Searching MAST for Our Star
Now that we know a little more about MAST, let's search it for any TPFs that may already exist for the star "TIC 408036613"...

```python
# target : the star we want to work with; the following naming formats are acceptable
    # name, e.g. "HAT-P-11"
    # identifying number, e.g. "TIC 408036613"
    # RA Dec coordinates, e.g. "285.67942179 +50.24130576" (decimal format) or "19:02:43.1 +50:14:28.7" (sexagesimal format)
    # an astropy.coordinates.SkyCoord object
search_result = lk.search_targetpixelfile(target="TIC 408036613")
search_result
```

In this scenario, no TPFs are available for download from MAST, meaning we must make our own from a FFI.
